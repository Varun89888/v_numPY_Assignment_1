### A1. Why are NumPY arrays generally preferred over plain Python lists for numerical computation in AI/ML workloads?


NumPy arrays are mainly used for numerical data and calculations. They can perform operations on many values at the same time, which is usually faster than processing values one by one with Python lists. NumPy also provides useful mathematical functions such as sum, mean, minimum, and maximum.

### A2. Explain vectorization. Why does replacing an explicit Python loop with a vectorized NumPy operation matter with large datasets?

Vectorization means performing an operation on many values at the same time instead of processing each value one by one using a loop. NumPy supports vectorized operations, which makes numerical calculations simpler and usually faster. This becomes especially useful when we work with large datasets because we can avoid writing explicit Python loops for every element.

### A3. For compatible 2D arrays A and B, explain A * B vs A @ B. Describe one neural-network situation for each.

A * B performs multiplication element by element, so values in the same positions are multiplied. A @ B performs matrix multiplication using rows and columns. In neural networks, * can be used for element-wise operations such as applying a mask, while @ is commonly used to multiply input features with weight matrices in a neural-network layer.

### A4. What is broadcasting in NumPy? State the general rule used to decide whether arrays of different shapes can be combined element-wise.

Broadcasting is a feature in NumPy that allows us to perform element-wise operations on arrays with different but compatible shapes. NumPy compares the dimensions starting from the right side. The dimensions are compatible when they are equal or when one of them is 1. Broadcasting is useful because it allows us to perform calculations without manually repeating or copying values.

### A5.Explain flatten() vs ravel(). In a memory-constrained application processing large images, which would you prefer and why?

Both flatten() and ravel() can convert an array into a one-dimensional array. The main difference is that flatten() creates a copy of the data, while ravel() tries to return a view of the original data when possible. When working with large images and limited memory, I would prefer ravel() because it can avoid creating an extra copy and therefore use less additional memory.

### A7. Why is it important to normalize or standardize numerical features before training most ML models? Explain normalization vs standardization.

Normalization and standardization help put numerical features on a similar scale before training a model. This can prevent features with large values from having too much influence. Normalization usually scales values to a fixed range such as 0 to 1. Standardization changes the values so that they have a mean of 0 and a standard deviation of 1.

Normalization - fixed range (usually 0 to 1)

Standardization - mean 0, standard deviation 1

### A8. Why do functions such as np.nanmean() exist? What happens if np.mean() is used directly on data containing NaN?

np.mean() calculates the average of all values, but if the array contains a NaN, the result will normally also be NaN.

np.nanmean() is useful when the data contains missing values represented by NaN. It calculates the mean while ignoring NaN values. In contrast, np.mean() normally returns NaN if any NaN value is present.

### C2. Write a function standardize(x) that takes a 1D NumPy array and returns its z-score standardized version using only NumPy.

```python
def standardize(x):
    mean = np.mean(x)
    std = np.std(x)
    return (x - mean) / std
x = np.array([10, 20, 30, 40, 50])
print(standardize(x))
```

Explanation: 
The function finds the mean and standard deviation of the array and uses them to convert each value into its standardized form.

### C4.Given X with shape (100, 5), compute column-wise mean and standard deviation and standardize the matrix in one vectorized expression.

```python
X = np.random.default_rng(42).random((100, 5))
mean = np.mean(X, axis=0)
std = np.std(X, axis=0)
X_standardized = (X - mean) / std
print(X_standardized.shape)
```

Explanation:

### C5. Implement y_pred = X @ w + b for a feature matrix X, weight vector w and bias b. Then compute MSE using y_true.

```python
X = np.array([[1, 2], [3, 4], [5, 6]])
w = np.array([0.5, 1.0])
b = 2
y_true = np.array([4, 7, 10])
y_pred = X @ w + b
mse = np.mean((y_true - y_pred) ** 2)
print("Predictions:", y_pred)
print("MSE:", mse)
```

Explanation: X @ w + b calculates the predicted values, and MSE calculates the average squared difference between the actual and predicted values.

### C7. Given a grayscale image as a 2D NumPy array with pixel values in [0, 255], normalize the values to [0, 1] using min-max normalization.

```python
image = np.array([
    [0, 128],
    [200, 255]
])
normalized = (image - image.min()) / (image.max() - image.min())
print(normalized)
```

Explanation:

# D1. Data Preprocessing Pipeline


Q. You are given a raw 1D or 2D NumPy array containing NaN values.​

Describe, step by step, the NumPy operations you would apply to:​

(a) detect missing values,​

(b) handle them appropriately and justify your strategy, and​

(c) normalize or standardize the clean data before it is fed to a model.​

Implement the full pipeline in code on a sample array of your own choosing, including at least one NaN.​



```python
import numpy as np

data = np.array([10, 20, np.nan, 40, 50])
print("Missing values:", np.isnan(data))
mean_value = np.nanmean(data)
clean_data = np.where(np.isnan(data), mean_value, data)
standardized = (clean_data - np.mean(clean_data)) / np.std(clean_data)
print("Clean data:", clean_data)
print("Standardized data:", standardized)
```

Explanation:

(a) Detect Missing Values

First, I use np.isnan() to check whether the array contains any missing (NaN) values.
np.isnan(data)

(b) Handle Missing Values

For numerical data, I can replace the missing value with the mean of that column. I use np.nanmean() because it calculates the mean without considering the NaN value.

(c) Standardize the Data

After filling the missing values, I standardize the data. Standardization changes the values so that they are on a similar scale.

The formula is:

Standardized value = (value - mean) / standard deviation

This can help the machine learning model work better when different features have different ranges.

```python
X = np.array([[1, 2],
              [2, 3],
              [3, 4]])

w = np.array([0.5, 1.0, 0.2])

b = 2.0

y_pred = X @ w + b

y_true = np.array([10, 20, 30])

mse = np.mean(y_true - y_pred ** 2)
```

Explonation:

There are two bugs in the code. 

First, X has shape (3, 2), so w must have 2 values, but the given w has 3 values. 

Second, the MSE formula is incorrect because y_pred ** 2 is calculated first due to operator precedence. 

We need to first subtract y_pred from y_true, then square the difference and calculate the mean

```python
#Corrected code
import numpy as np

X = np.array([[1, 2],
              [2, 3],
              [3, 4]])

w = np.array([0.5, 1.0])
b = 2.0
y_pred = X @ w + b
y_true = np.array([10, 20, 30])
mse = np.mean((y_true - y_pred) ** 2)
print("Predictions:", y_pred)
print("MSE:", mse)
```

Explanation

X contains 3 data samples and 2 features, so w contains 2 weights. The @ operator calculates the weighted values, and b is added as the bias. Finally, MSE is calculated by finding the difference between actual and predicted values, squaring the differences, and taking their mean.

E1. Per-Column Min–Max Normalization

Implement min-max normalization for a 2D feature matrix on a per-column (per-feature) basis, rather than using a single global 
min/max for the whole array. Your solution must be fully vectorized, with no explicit loops over rows or columns.​

```python
import numpy as np

X = np.array([
    [10, 100],
    [20, 200],
    [30, 300]
])
X_min = np.min(X, axis=0)
X_max = np.max(X, axis=0)
X_normalized = (X - X_min) / (X_max - X_min)
print("Column minimums:", X_min)
print("Column maximums:", X_max)
print("Normalized matrix:")
print(X_normalized)
```

Explanation:

Here, each column is normalized separately using its own minimum and maximum value. axis=0 finds the minimum and maximum for each feature column. Broadcasting allows these values to be used for all rows without using loops. The normalized values are between 0 and 1.

# E2. Three-Level Score Classification

Given a 1D array of exam scores, use a single vectorized expression (no if/else statements and no explicit loops) to classify 
each score as:​

Fail (< 40), Average (40–70 inclusive), or Excellent (> 70).​

Hint: You may need to nest or combine np.where() calls.​

```python
import numpy as np

scores = np.array([25, 40, 55, 70, 85, 95])
result = np.where(
    scores < 40,
    "Fail",
    np.where(scores <= 70, "Average", "Excellent")
)
print("Scores:", scores)
print("Classification:", result)
```

Explanation:

I used nested np.where() to classify all scores without using loops or if/else statements. Scores below 40 are classified as Fail, scores from 40 to 70 are classified as Average, and scores above 70 are classified as Excellent. Because np.where() works on the complete array at once, the solution is vectorized.
